Plot these points and draw a curve y = (a)(xloglogx) + b with arbitrary a and b to best fit the points. scale the x and y axis the same
Curve 1 (yellow): (1000001, 25), (10000001, 340), (100000001, 320), (1000000001, 5200), (2000000001, 11900)
Curve 2 (blue): (1000001, 1), (10000001, 10), (100000001, 310), (1000000001, 4250), (2000000001, 10600)
Curve 3 (green): (1000001, 1), (10000001, 20), (100000001, 300), (1000000001, 4800), (2000000001, 10550)

```python
# TODO scale better for clearer depiction of my experiment

import numpy as np
import matplotlib.pyplot as plt
from math import log
from scipy.optimize import curve_fit

# --- Model function: y = a * (x log log x) + b ---
def model(x, a, b):
    return a * (x * np.log(np.log(x))) + b

# --- Data ---
curve1_x = np.array([1_000_001, 10_000_001, 100_000_001, 1_000_000_001])
curve1_y = np.array([25, 340, 320, 5200])

curve2_x = np.array([1_000_001, 10_000_001, 100_000_001, 1_000_000_001])
curve2_y = np.array([10, 50, 650, 7550])

curve3_x = np.array([1_000_001, 10_000_001, 100_000_001, 1_000_000_001])
curve3_y = np.array([10, 70, 635, 8200])

# Fit parameters for each curve
popt1, _ = curve_fit(model, curve1_x, curve1_y)
popt2, _ = curve_fit(model, curve2_x, curve2_y)
popt3, _ = curve_fit(model, curve3_x, curve3_y)

# Generate smooth x for drawing fitted curves
x_smooth = np.linspace(1_000_001, 1_000_000_001, 500)

# --- Plot ---
fig, ax = plt.subplots()

# Raw points
ax.scatter(curve1_x, curve1_y, color='yellow', label='Curve 1 Points')
ax.scatter(curve2_x, curve2_y, color='blue', label='Curve 2 Points')
ax.scatter(curve3_x, curve3_y, color='green', label='Curve 3 Points')

# Fitted curves
ax.plot(x_smooth, model(x_smooth, *popt1), color='yellow', linestyle='--', label='Curve 1 Fit')
ax.plot(x_smooth, model(x_smooth, *popt2), color='blue', linestyle='--', label='Curve 2 Fit')
ax.plot(x_smooth, model(x_smooth, *popt3), color='green', linestyle='--', label='Curve 3 Fit')

# Equal scaling
ax.set_aspect('equal', adjustable='datalim')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Curve Fit: y = a(x log log x) + b')
ax.legend()

# scale axis
plt.xscale("log")
plt.yscale("log")

plt.grid(True)
plt.show()

# Print fitted parameters
print("Curve 1 fit parameters (a, b):", popt1)
print("Curve 2 fit parameters (a, b):", popt2)
print("Curve 3 fit parameters (a, b):", popt3)
```
